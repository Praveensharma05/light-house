```python
import numpy as np
arr = np.array([[1,2,3],[4,5,6],[6,7,8]])
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.ndim)

#int to float

float_arr = arr.astype(np.float64)
print(float_arr)
```

```python
#reshaping of array
arr = np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)

reshaped = arr.reshape((3,2))
print(reshaped, reshaped.shape)
flattened = arr.flatten()
print(flattened, flattened.shape)
```

```python
#slicing in numpy

arr = np.array([1,2,4,5,6,7])
print(arr)
sub_arr = arr[1:3]
print(sub_arr)

sub_arr[0] = 100
print(sub_arr)
print(arr)

#
```

```python
#normalisation

arr = np.array([[1,2,3],[4,5,6]])
print(arr)
mean = np.mean(arr)
print(mean)
std = np.std(arr)
print(std)

normalized_arr = (arr - mean) /std
print(normalized_arr)
mean = np.mean(normalized_arr)
print(mean)
std = np.std(normalized_arr)
print(std)



```
